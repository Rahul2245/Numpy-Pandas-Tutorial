```python
bios_new['height_category']=bios['height_cm'].apply(lambda x:'Short' if x<165 else ('Average' if x<185  else 'Tall'))
bios_new
```

```python
def categorize_athelete(row):
    if row['height_cm']<175 and row['weight_kg']<70:
        return 'Lightweight'
    elif row['height_cm']<185 or row['weight_kg']<=80:
        return 'Middleweight'
    else:
        return 'heavyweight'
bios_new['category']=bios.apply(categorize_athelete,axis=1); 
bios_new   
```

```python
nocs = pd.read_csv('./complete-pandas-tutorial/data/noc_regions.csv')
```

```python
bios_up=pd.merge(bios_new,nocs,left_on='born_country',right_on='NOC',how='left')
bios_up
```
