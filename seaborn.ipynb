```python
#heatmap
columns = [
    "bill_length_mm",
    "bill_depth_mm",
    "flipper_length_mm",
    "body_mass_g"
]
corr=penguins[columns].corr()
# plt.figure(figsize=(8,6))

sns.heatmap(
    data=corr,
    annot=True,        # show values
    cmap="coolwarm",   # color theme
    linewidths=1,
    # fmt=".2f"
    vmin=-0.2
)

# plt.title("Penguins Correlation Heatmap")
# plt.show()
```

```python
#rug plot
sns.rugplot(
    data=penguins,
    x="body_mass_g",
    height=0.4
)
```

```python
sns.pairplot(
    data=penguins,
    hue="species"
)
```

```python
#pairgrid

g = sns.PairGrid(
    data=penguins,
    hue="species"
)

# Upper triangle
g.map_upper(sns.scatterplot)

# Lower triangle
g.map_lower(sns.kdeplot)

# Diagonal
g.map_diag(sns.histplot)

# Legend
g.add_legend()
```
