```python
import seaborn as sns
import pandas as pd
import numpy as np
```

```python


sns.get_dataset_names()
```

```python
penguins=sns.load_dataset('penguins')
penguins.head()
```

```python
penguins["species"].value_counts()
```

```python
penguins["island"].value_counts()
```

```python
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="sex" )
```

```python
sns.set_style("darkgrid")
#white
#dark
#whitegrid
#darkgrid
#ticks
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island" )
```

```python
sns.set_style("darkgrid")
sns.set_context("poster")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island" ,palette="Dark2")
```

```python
sns.set_style("darkgrid")
# sns.set_context("poster")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island" ,palette="Dark2",style="sex")
```

```python
sns.scatterplot(data=penguins,x="species",y="body_mass_g",hue="island",style="sex",alpha=True)
```

```python
sns.stripplot(data=penguins,x="species",y="body_mass_g",hue="island",dodge=True,jitter=True)
```

```python
sns.set_context("talk")
sns.swarmplot(data=penguins,x="species",y="body_mass_g",hue="island")
```

```python
#histogram
sns.set_context("notebook")
sns.histplot(data=penguins,x="body_mass_g",hue="sex",multiple="stack")
```

```python
sns.regplot(data=penguins, x="body_mass_g",y="flipper_length_mm")
```

```python
sns.regplot(data=penguins, x="body_mass_g",y="flipper_length_mm",scatter=False)
```

```python
#line plot
sns.set_style("whitegrid")
sns.lineplot(data=penguins,x="body_mass_g",y="flipper_length_mm",hue="island",style="sex")
```

```python
#joint plot
sns.jointplot(data=penguins,x="body_mass_g",y="flipper_length_mm")
```

```python
sns.jointplot(data=penguins,x="body_mass_g",y="flipper_length_mm",hue="sex")
```

```python
sns.jointplot(data=penguins,x="body_mass_g",y="flipper_length_mm",hue="island",kind="kde")
```

```python
sns.jointplot(data=penguins,x="body_mass_g",y="flipper_length_mm",hue="island",kind="scatter")
```

```python
sns.jointplot(data=penguins,x="body_mass_g",y="flipper_length_mm",kind="reg")
```

```python
sns.barplot(data=penguins,x="species",y="body_mass_g",hue="sex",estimator=np.var)
```
