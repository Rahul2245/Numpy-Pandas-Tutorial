```python
data = {
    "salary": [
        22000, 24000, 25000, 26000, 27000,
        28000, 29000, 30000, 31000, 32000,
        33000, 34000, 35000, 36000, 37000,
        38000, 39000, 40000, 42000, 45000,
        47000, 48000, 50000, 52000, 55000,
        58000, 60000, 62000, 65000, 70000,
        75000, 80000, 85000, 90000, 95000,
        100000
    ]
}
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
#line plot

plt.plot(df["salary"],color="red",marker="o",linestyle=":" ,linewidth="2" )
plt.grid()
plt.show()
```

```python
#histogram
plt.hist(df["salary"],bins=5,color="green")
plt.show()
```

```python
#boxplot
plt.boxplot(df["salary"])
plt.show()
```

```python
df.count()
plt.boxplot(df["salary"])
plt.show()
```

```python
df.loc[20]=[0]
plt.boxplot(df["salary"])
plt.show()
```
