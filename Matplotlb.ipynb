```python
df["Age"] = [
    22, 24, 25, 23, 26,
    27, 28, 29, 30, 24,
    25, 31, 32, 26, 27,
    28, 29, 33, 34, 35,
    23, 24, 26, 27, 28,
    29, 30, 31, 32, 33,
    25, 26, 27, 28, 29
]
df.head()
```

```python
#scatter plot
plt.scatter(df["Age"],df["salary"])
plt.show()
```

```python
#line plot
sort_age=df.sort_values("Age")
plt.plot(sort_age["Age"],df["salary"],color="red",marker="*",linestyle=":",linewidth="2")
plt.grid()
plt.show()
```

```python
plt.bar(sort_age["Age"],df["salary"],color="green",align="edge")
plt.show()
```

```python
#bivariate : Numerical-categorial
hr_sal = df[df["dept"] == "HR"]["salary"]
it_sal = df[df["dept"] == "IT"]["salary"]
finance_sal = df[df["dept"] == "Finance"]["salary"]
marketing_sal = df[df["dept"] == "Marketing"]["salary"]
sales_sal = df[df["dept"] == "Sales"]["salary"]
```

```python
#boxplot
plt.boxplot( [hr_sal,it_sal,finance_sal,marketing_sal,sales_sal],labels=["Hr","it","finance","marketing","sales"])
plt.show()
```

```python
#pie chart
salary_bydept=df.groupby("dept")["salary"].sum()
salary_bydept
```

```python
plt.pie(salary_bydept,labels=salary_bydept.index,autopct="%1.0f",shadow=True)
plt.axis("equal")
plt.show()
```

```python
hr_mean = sum(hr_sal) / len(hr_sal)
it_mean = sum(it_sal) / len(it_sal)
finance_mean = sum(finance_sal) / len(finance_sal)
marketing_mean = sum(marketing_sal) / len(marketing_sal)
sales_mean = sum(sales_sal) / len(sales_sal)
```

```python
plt.bar(["Hr","it","finance","marketing","sales"],[hr_mean,it_mean,finance_mean,marketing_mean,sales_mean],color="green")
plt.grid()
plt.show()
```

```python
#multivariate analysis: 3numerical
df["experience"] = [
    0.5, 1, 1, 1.5, 2,
    2, 2.5, 3, 3, 3.5,
    4, 4.5, 5, 5.5, 6,
    6.5, 7, 7.5, 8, 8.5,
    9, 9.5, 10, 11, 12,
    13, 14, 15, 16, 17,
    18, 20, 22, 24, 26
]
df.head()
```

```python
#bubble plot
plt.scatter(df["Age"],df["salary"],s=df["experience"]*20,color="skyblue")
plt.title("Age vs Salary vs Experience")
plt.xlabel( "Age")
plt.ylabel("Salary")
plt.show()
```

```python
#numerical-2
#categorial-1
color_map = {
    "HR": "red",
    "IT": "blue",
    "Finance": "green",
    "Sales": "orange",
    "Marketing":"black"
}
plt.scatter(df["Age"],df["salary"],c=df["dept"].map(color_map).fillna("gray"))
plt.title("Age vs Salary vs dept")
plt.xlabel( "Age")
plt.ylabel("Salary")
plt.show()
 
```

```python
color_map = {
    "HR": "red",
    "IT": "blue",
    "Finance": "green",
    "Sales": "orange",
    "Marketing": "black"
}

for dept, color in color_map.items():
    
    subset = df[df["dept"] == dept]
    
    plt.scatter(
        subset["Age"],
        subset["salary"],
        color=color,
        label=dept
    )

plt.title("Age vs Salary vs Dept")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.legend()

plt.show()
```
